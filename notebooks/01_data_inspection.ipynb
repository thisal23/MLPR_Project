Import libraries and configure paths

```python
from pathlib import Path
import pandas as pd

# Locate project directory
PROJECT_ROOT = Path.cwd()

# Dataset path
DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "paddy_production.csv"
)

print("Project directory:", PROJECT_ROOT)
print("Dataset path:", DATA_PATH)

if not DATA_PATH.exists():
    raise FileNotFoundError(f"Dataset not found: {DATA_PATH}")
```

Load the dataset

```python
print("Loading paddy production dataset...")

df = pd.read_csv(DATA_PATH)

# Remove accidental spaces in column names
df.columns = df.columns.str.strip()

print("Dataset loaded successfully!")
```

Dataset Overview

```python
print("========== DATASET OVERVIEW ==========")

print("Number of records:", len(df))
print("Number of columns:", len(df.columns))

print("\nColumn names:")
print(df.columns.tolist())
```

Display records

```python
print("First five records:")
display(df.head())

print("Last five records:")
display(df.tail())
```

Data types and Missing values

```python
print("Data types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum())
```

Duplicate records

```python
print("Duplicate records:", df.duplicated().sum())
```

Statistical summary

```python
print("Statistical summary:")
display(df.describe())
```

Year range

```python
print("Year range:")
print(df["Year"].min(), "-", df["Year"].max())
```

Production statistics

```python
print("Production statistics:")
display(df["Production 000 Mt."].describe())
```

Check missing years

```python
years = sorted(df["Year"].tolist())

missing_years = [
    year
    for year in range(years[0], years[-1] + 1)
    if year not in years
]

print("Missing years:", missing_years)
```
